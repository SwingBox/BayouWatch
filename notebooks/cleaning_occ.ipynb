```python
# nettoyage eventDate 2#
df_occ_c_reordered['eventDate'] = df_occ_c_reordered['eventDate'].str.replace(r'/.*', '', regex=True)
```

```python
# fonction avec regex pour nettoyer scientificName" #
def nettoyer_nom_scientifique(nom):
    if pd.isna(nom):
        return nom  # Garde les NaN

    # Supprimer les guillemets
    nom = nom.replace('"', '').replace("'", '')

    # Supprimer les contenus entre parenthèses (auteur, dates, etc.)
    nom = re.sub(r"\(.*?\)", "", nom)

    # Supprimer les abréviations type "L.", "Hüb." (majuscule + point)
    nom = re.sub(r"\b[A-Z]\.", "", nom)

    # Supprimer les années (4 chiffres consécutifs)
    nom = re.sub(r"\b\d{4}\b", "", nom)

    # Supprimer les virgules, points superflus, ou caractères non alphabétiques isolés
    nom = re.sub(r"[^a-zA-Z\s\-]", "", nom)

    # Réduire les espaces multiples à un seul
    nom = re.sub(r"\s+", " ", nom)

    # Enlever les espaces en début/fin
    nom = nom.strip()

    return nom
```

```python
df_occ_c_reordered['scientificName'] = df_occ_c_reordered['scientificName'].apply(nettoyer_nom_scientifique)
```

```python
for _, rows in df_occ_c_reordered.iterrows():
    if rows["species"] == np.nan:
        rows["species"] = rows["scientificName"]

# Fonctionnel mais plus long
```
