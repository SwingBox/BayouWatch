```python
import geopandas as gpd

# Charge le fichier GeoJSON
gdf = gpd.read_file("map.geojson")

# Récupère le polygone (la zone de "Louisiane-Bayou")
polygon = gdf.geometry.iloc[0]
```

```python
import pandas as pd

df_occ_c_reordered = pd.read_csv("df_occ_c_reordered.csv")
```

```python
from shapely.geometry import Point

gdf_occ_c_reordered = gpd.GeoDataFrame(
    df_occ_c_reordered,
    geometry=gpd.points_from_xy(df_occ_c_reordered['decimalLongitude'], df_occ_c_reordered['decimalLatitude']),
    crs="EPSG:4326")

gdf_occ_c_reordered_filtered = gdf_occ_c_reordered.sjoin(gpd.GeoDataFrame(geometry=[polygon], crs="EPSG:4326"), predicate='within', how='inner')
```

```python
gdf_occ_c_reorderedn_final = gdf_occ_c_reordered_filtered.drop(columns='geometry')
```
